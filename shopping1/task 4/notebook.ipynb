# Catalog Expansion Decision Memo

The VP of Supply Chain has requested a decision memo evaluating whether Produce should receive the next catalog expansion budget.

```python
from IPython.display import HTML
import pandas as pd


DATA_DIR = "../../data/shop"

products = pd.read_csv(
    f"{DATA_DIR}/products.csv"
)

departments = pd.read_csv(
    f"{DATA_DIR}/departments.csv"
)

aisles = pd.read_csv(
    f"{DATA_DIR}/aisles.csv"
)

products.head()
```

## Calculate catalog summary statistics

Calculate the number of departments and aisles and the average number of products per aisle.

```python
num_departments = len(departments)
num_aisles = len(aisles)

avg_products_per_aisle = (
    products
    .groupby("aisle_id")
    .size()
    .mean()
)

print(
    "Departments:",
    num_departments,
)

print(
    "Aisles:",
    num_aisles,
)

print(
    "Average products per aisle:",
    f"{avg_products_per_aisle:.2f}",
)
```

## Rank departments by product assortment

Count products within each department, rank departments by assortment size, and determine Produce's position.

```python
department_counts = (
    products
    .groupby("department_id")
    .size()
    .rename("product_count")
    .reset_index()
    .merge(
        departments,
        on="department_id",
    )
    .sort_values(
        "product_count",
        ascending=False,
    )
    .reset_index(drop=True)
)

department_counts["rank"] = (
    department_counts.index + 1
)

produce = department_counts.query(
    "department == 'produce'"
).iloc[0]

top5 = department_counts.head(5)


print(
    top5[
        ["rank", "department", "product_count"]
    ].to_string(index=False)
)

print(
    "\nProduce rank:",
    produce["rank"],
)
```

## Evaluate the proposal

Use Produce's computed rank and the leading department to determine whether the catalog evidence supports the director's rationale.

```python
recommendation = (
    f"The catalog evidence does not support the director's rationale. " "\n" 
    f"Produce ranks {produce['rank']} by product count, while " "\n" 
    f"{department_counts.iloc[0]['department']} has the largest assortment. " "\n" 
    f"Catalog breadth alone therefore does not justify prioritizing Produce."
)

print(recommendation)
```
