# Catalog Expansion Decision Memo

The VP of Supply Chain has requested a decision memo evaluating whether Produce should receive the next catalog expansion budget.

The analysis summarizes the structure of the Instacart product catalog and evaluates whether the computed statistics support the director's rationale.

```python
from IPython.display import HTML
import pandas as pd


DATA_DIR = "../../data/shop"

products = pd.read_csv(
    f"{DATA_DIR}/products.csv"
)

departments = pd.read_csv(
    f"{DATA_DIR}/departments.csv"
)

aisles = pd.read_csv(
    f"{DATA_DIR}/aisles.csv"
)


# Department product counts and ranking.
department_counts = (
    products
    .groupby("department_id")
    .size()
    .rename("product_count")
    .reset_index()
    .merge(
        departments,
        on="department_id",
    )
    .sort_values(
        "product_count",
        ascending=False,
    )
    .reset_index(drop=True)
)

department_counts["rank"] = (
    department_counts.index + 1
)

produce = department_counts.query(
    "department == 'produce'"
).iloc[0]

top5 = department_counts.head(5)


# Catalog statistics.
num_departments = len(departments)
num_aisles = len(aisles)

avg_products_per_aisle = (
    products
    .groupby("aisle_id")
    .size()
    .mean()
)


recommendation = (
    f"The catalog evidence does not support the director's rationale. "
    f"Produce ranks {produce['rank']} by product count, while "
    f"{department_counts.iloc[0]['department']} has the largest assortment. "
    f"Catalog breadth alone therefore does not justify prioritizing Produce."
)

HTML(
    f"""
<div style="
    background:#f5f5f5;
    padding:24px;
    border-radius:12px;
    font-family:Georgia,serif;
">

<h2 style="
    background:#2f3b52;
    color:white;
    padding:12px;
    border-radius:8px;
">
Supply Chain Decision Memo
</h2>

<h3>Catalog Summary</h3>

<ul>
    <li>Departments: {num_departments}</li>
    <li>Aisles: {num_aisles}</li>
    <li>Average products per aisle:
        {avg_products_per_aisle:.2f}</li>
    <li>Produce rank:
        {produce["rank"]}</li>
</ul>

<h3>Top Five Departments</h3>

{top5[["rank", "department", "product_count"]]
    .to_html(index=False)}

<h3>Recommendation</h3>

<p>{recommendation}</p>

<p style="
    color:gray;
    font-style:italic;
">
Source: Instacart product catalog
</p>

</div>
"""
)
```

## Conclusion

The computed catalog statistics do not support the claim that Produce has the broadest product assortment.

Produce ranks below several other departments in unique product count, indicating that catalog breadth alone does not justify prioritizing Produce for the next catalog expansion budget.

The recommendation is therefore based directly on the statistics computed from the dataset rather than on the initial proposal.
